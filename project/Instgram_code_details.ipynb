```python
with open("initialdata.txt",encoding="utf-8") as fl :
    data = fl.read()
```

```python
chunks = data.split("\n\n")
chunks = [c for c in chunks if len(c)>3]
```

```python
def parse_chunks(chunk) :
    chunk = chunk.strip()
    sep_chunks = chunk.split("\n")

    #User Name
    user_name = sep_chunks[0]
    
    #posts Finding
    no_of_post = sep_chunks[1]
    
    no_of_post = int(no_of_post.split(" post")[0].replace(",",""))
    print(no_of_post)
    
    # Followers Finding
    no_of_followers = sep_chunks[2]
    no_of_followers = float(no_of_followers.split(" followers")[0].replace(",","").replace("K","").replace("M",""))
    if("K" in sep_chunks[2]):
        no_of_followers = int(no_of_followers * 1000)
    elif("M" in sep_chunks[2]) :
        no_of_followers = int(no_of_followers * 1000000)
    else :
        no_of_followers = no_of_followers
    
    # Following Finding
    no_of_following = sep_chunks[3]
    no_of_following = float(no_of_following.split(" following")[0].replace(",","").replace("K","").replace("M",""))
    if("K" in sep_chunks[3]) :
        no_of_following = int(no_of_following * 1000)
    elif("M" in sep_chunks[2]) :
       no_of_following = int(no_of_following * 1000000)
    else :
        no_of_following = no_of_following
    
    # Name (Account/person Name)
    name = sep_chunks[4]
    
    if (len(sep_chunks)>5) :
        type_of_page = sep_chunks[5]           # Categories Finding(type of page)
        bio = "\n".join(sep_chunks[6:])                     # Bio Finding
    else :
        type_of_page = "Unknown"
        bio = ""
    return {
          "user_name" : user_name, 
          "no_of_post" : no_of_post, 
          "no_of_followers" : no_of_followers, 
          "no_of_following" : no_of_following ,
          "name" : name , 
          "type_of_page": type_of_page,
          "bio" : bio
         }
```

```python
# Number Of Posts 
```

```python
all_chunks = []
for chunk in chunks :
    parsed_chunk = parse_chunks(chunk)
    all_chunks.append(parsed_chunk)
```

```python
import json 
s = json.dumps(all_chunks, indent= 4)
with open("insta_output_.json", "w") as fl :
    fl.write(s)
print("Instagram Data (After The Parsing ) Saved in insta_output_.json,plz check")
```

# Who has maximum Posts :
